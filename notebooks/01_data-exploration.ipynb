# Data Exploration


```python
import pandas as pd

from credit_risk.config import RAW_DATA_DIR
```

```python
credit_risk_df = pd.read_csv(RAW_DATA_DIR / "application_train.csv")
```

```python
n_rows, n_cols = credit_risk_df.shape

print(f"Number of rows: {n_rows:,}\nNumber of columns: {n_cols}")
```

### Available Columns

```python
column_names = credit_risk_df.columns.tolist()
n_display_cols = 4

columns_df = pd.DataFrame({
    i: pd.Series(column_names[i::n_display_cols])
    for i in range(n_display_cols)
})

print(columns_df.to_string(index=False, header=False, na_rep=""))
```

### Categorical Columns

```python
categorical_cols = credit_risk_df.select_dtypes(include=["category", "object", "string"]).columns.tolist()

print(f"Categorical columns: {categorical_cols}")
```

### Description of Selected Columns

Selected variables from `application_train.csv`:

| Column | Description |
|---|---|
| `SK_ID_CURR` | Unique loan application identifier. |
| `TARGET` | Binary target indicating repayment difficulties (1) or no recorded difficulties (0). |
| `AMT_INCOME_TOTAL` | Applicant's total income. |
| `AMT_CREDIT` | Credit amount of the loan. |
| `AMT_ANNUITY` | Loan annuity amount. |
| `AMT_GOODS_PRICE` | Price of the goods financed by the loan. |
| `DAYS_BIRTH` | Applicant's age in days relative to the application date (negative values). |
| `DAYS_EMPLOYED` | Days since the applicant started their current employment, relative to the application date. |
| `NAME_INCOME_TYPE` | Applicant's income source or employment category. |
| `NAME_EDUCATION_TYPE` | Applicant's highest education level. |
| `EXT_SOURCE_1`, `EXT_SOURCE_2`, `EXT_SOURCE_3` | Normalized scores from external data sources. |
| `CODE_GENDER` | Applicant's gender. |
| `CNT_CHILDREN` | Number of children of the applicant. |
| `CNT_FAM_MEMBERS` | Number of family members in the applicant's household. |
| `NAME_FAMILY_STATUS` | Applicant's family status. |
| `NAME_HOUSING_TYPE` | Applicant's housing situation. |
| `OCCUPATION_TYPE` | Applicant's occupation. |
| `FLAG_OWN_CAR` | Indicates whether the applicant owns a car. |
| `FLAG_OWN_REALTY` | Indicates whether the applicant owns a house or flat. |
| `NAME_CONTRACT_TYPE` | Type of loan contract (cash or revolving). |
| `REGION_RATING_CLIENT` | Rating of the applicant's residential region. |

Source: [Home Credit Default Risk — Kaggle](https://www.kaggle.com/competitions/home-credit-default-risk/data). Column descriptions are summarized and paraphrased from the dataset documentation (`HomeCredit_columns_description.csv`).



### Target Variable

```python
default_rate = credit_risk_df["TARGET"].mean()

print(f"Default rate: {default_rate:.2%}")
```

### Missing Values

```python
missing_counts = credit_risk_df.isna().sum()

max_missing_count = missing_counts.max()

if max_missing_count > 0:
    most_missing_columns =  [column for column in credit_risk_df.columns if missing_counts[column] == max_missing_count]
    print(f"Columns with the highest number of NAs ({max_missing_count:,}): {most_missing_columns}")
else:
    print("No columns has NAs")
```

### SK ID (Loan Application Identifier)

```python
is_unique_sk_id_curr = credit_risk_df["SK_ID_CURR"].is_unique

print(f"SK_ID_CURR is {'unique' if is_unique_sk_id_curr else 'not unique'} for every observation.")
```
